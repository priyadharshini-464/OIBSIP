```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
%matplotlib inline
```

```python
df = pd.read_csv('/content/retail_sales_dataset.csv')
df.head()
```

```python
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
df = df.drop_duplicates()
```

```python
df['year'] = df['Date'].dt.year
df['month'] = df['Date'].dt.month
df['month_name'] = df['Date'].dt.strftime('%b')
df['weekday'] = df['Date'].dt.day_name()
```

```python
monthly_sales = df.groupby(df['Date'].dt.to_period('M'))['Total Amount'].sum()

plt.figure(figsize=(12,5))
monthly_sales.plot(kind='line')
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales Amount")
plt.show()
```

```python
cat_sales = df.groupby('Product Category')['Total Amount'].sum().reset_index()

plt.figure(figsize=(8,5))
sns.barplot(data=cat_sales, x='Product Category', y='Total Amount')
plt.title("Total Sales by Product Category")
plt.xticks(rotation=45)
plt.show()
```

```python
gender_sales = df.groupby('Gender')['Total Amount'].sum().reset_index()

plt.figure(figsize=(6,4))
sns.barplot(data=gender_sales, x='Gender', y='Total Amount')
plt.title("Total Spending by Gender")
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.histplot(df['Age'], kde=True, bins=20)
plt.title("Customer Age Distribution")
plt.show()
```

```python
qty_sales = df.groupby('Product Category')['Quantity'].sum().reset_index()

plt.figure(figsize=(8,5))
sns.barplot(data=qty_sales, x='Product Category', y='Quantity')
plt.title("Quantity Sold per Category")
plt.xticks(rotation=45)
plt.show()
```

🔍 Insights & Recommendations

1. Monthly Sales Trend
Sales show consistent patterns across months, with slight peaks indicating seasonal demand.
This suggests planning inventory ahead of high-demand months.

2. Product Category Performance

Electronics generates high revenue.

Clothing sells in high quantities but lower revenue, which means lower price margin.

Beauty category has stable medium sales.

3. Gender-Based Spending

Female customers spend more overall compared to males.

Indicates female-focused promotions will be more effective.

4. Age Group Insights

Highest spending is from customers aged 30–45.

Marketing campaigns can target mid-age working professionals.

5. Quantity vs Revenue

Clothing → highest quantity sold

Electronics → fewer products sold but very high revenue
This indicates product pricing strategy differences.

6. Recommendations

Increase stock of Electronics before peak months.

Boost offers for Beauty and Clothing categories to increase revenue.

Target 25–45 age group with ads and promotions.

Female customers can be targeted with loyalty programs.
